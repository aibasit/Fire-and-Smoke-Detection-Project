# 02 - Train YOLOv8 Fire and Smoke Model

Train on a laptop/desktop/Colab when possible. Jetson Nano should mainly be used for inference.

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_YAML = PROJECT_ROOT / 'datasets' / 'fire_smoke_roboflow_v4' / 'data_project.yaml'
RUNS_DIR = PROJECT_ROOT / 'runs'
print(DATA_YAML)
print(DATA_YAML.exists())
```

```python
# Run this if ultralytics is not installed in your current notebook environment.
# %pip install ultralytics
```

```python
import ultralytics
from ultralytics import YOLO

print('Ultralytics version:', ultralytics.__version__)
ultralytics.checks()
```

For Jetson Nano deployment, start with `yolov8n.pt`. If accuracy is not enough and speed is acceptable, try `yolov8s.pt` later.

```python
model = YOLO('yolov8n.pt')

results = model.train(
    data=str(DATA_YAML),
    epochs=50,
    imgsz=640,
    batch=16,
    workers=4,
    project=str(RUNS_DIR),
    name='fire_smoke_yolov8n',
    exist_ok=True,
    patience=15,
    pretrained=True,
    optimizer='auto'
)

print(results)
```

```python
best_model = RUNS_DIR / 'fire_smoke_yolov8n' / 'weights' / 'best.pt'
last_model = RUNS_DIR / 'fire_smoke_yolov8n' / 'weights' / 'last.pt'
print('Best model:', best_model)
print('Exists:', best_model.exists())
print('Last model:', last_model)
print('Exists:', last_model.exists())
```
