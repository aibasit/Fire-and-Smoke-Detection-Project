# 04 - Export Model for Jetson Nano

Export the trained YOLOv8 model. Start with ONNX because it is portable. TensorRT export is best done directly on the Jetson Nano.

````python
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

BEST_MODEL = PROJECT_ROOT / 'runs' / 'fire_smoke_yolov8n' / 'weights' / 'best.pt'
EXPORT_DIR = PROJECT_ROOT / 'models' / 'exported'
EXPORT_DIR.mkdir(parents=True, exist_ok=True)

print('Best model:', BEST_MODEL)
print('Exists:', BEST_MODEL.exists())
````

````python
from ultralytics import YOLO

model = YOLO(str(BEST_MODEL))
onnx_path = model.export(format='onnx', imgsz=640, simplify=True, opset=12)
print('ONNX export:', onnx_path)
````

For Jetson Nano, first test `best.pt` with Ultralytics. If FPS is too low, export TensorRT on the Jetson itself:

```python
from ultralytics import YOLO
model = YOLO('best.pt')
model.export(format='engine', imgsz=640, half=True)
```

TensorRT export depends on JetPack, CUDA, TensorRT, and Python versions, so it is more reliable on the actual Jetson.
