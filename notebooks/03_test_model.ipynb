# 03 - Evaluate and Test the Trained Model

Use this after training. It validates the model and runs sample predictions on test images.

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

DATASET_DIR = PROJECT_ROOT / 'datasets' / 'fire_smoke_roboflow_v4'
DATA_YAML = DATASET_DIR / 'data_project.yaml'
BEST_MODEL = PROJECT_ROOT / 'runs' / 'fire_smoke_yolov8n' / 'weights' / 'best.pt'

print('Best model:', BEST_MODEL)
print('Exists:', BEST_MODEL.exists())
```

```python
from ultralytics import YOLO

model = YOLO(str(BEST_MODEL))
metrics = model.val(data=str(DATA_YAML), split='test', imgsz=640, conf=0.25)
print(metrics)
```

```python
import random

test_images = list((DATASET_DIR / 'test' / 'images').glob('*'))
sample_images = random.sample(test_images, k=min(8, len(test_images)))

predictions = model.predict(
    source=[str(p) for p in sample_images],
    imgsz=640,
    conf=0.35,
    save=True,
    project=str(PROJECT_ROOT / 'runs'),
    name='sample_predictions',
    exist_ok=True
)

print('Saved predictions to:', PROJECT_ROOT / 'runs' / 'sample_predictions')
```
